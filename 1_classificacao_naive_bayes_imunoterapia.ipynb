# Classificação com Naive Bayes - Imunoterapia (UCI)
Mesmo tipo de análise do projeto de crioterapia (GaussianNB), aplicado ao dataset irmão **Immunotherapy** do UCI.

Dataset: https://archive.ics.uci.edu/dataset/428/immunotherapy
Baixe o arquivo `Immunotherapy.xlsx` e coloque na mesma pasta deste notebook.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Carrega o dataset (arquivo baixado do UCI)
df = pd.read_excel("Immunotherapy.xlsx")
df.head()
```

```python
# Visão geral dos dados
df.describe().round(2)
```

```python
# Balanceamento das classes (resultado do tratamento)
sns.countplot(x="Result_of_Treatment", data=df, palette="Set2")
plt.title("Resultado do Tratamento (0 = Negativo, 1 = Positivo)")
plt.show()
```

```python
# Correlação entre as variáveis
plt.figure(figsize=(7,5))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="RdBu_r")
plt.title("Mapa de Correlação")
plt.show()
```

```python
# Separa variáveis preditoras (X) e alvo (y)
X = df.drop(columns=["Result_of_Treatment"])
y = df["Result_of_Treatment"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
# Treina o modelo Naive Bayes (mesmo algoritmo do app de crioterapia)
modelo_NB = GaussianNB()
modelo_NB.fit(X_train, y_train)

y_pred = modelo_NB.predict(X_test)

print("Acurácia:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
# Matriz de confusão
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
plt.xlabel("Predito")
plt.ylabel("Real")
plt.title("Matriz de Confusão")
plt.show()
```

```python
# Classificação de um novo paciente, com probabilidade (igual ao bottle_app_crioterapia.py)
novo_paciente = [[1, 25, 6, 3, 1, 80, 20]]  # sex, age, Time, Number_of_Warts, Type, Area, induration_diameter

res = modelo_NB.predict(novo_paciente)
prob = modelo_NB.predict_proba(novo_paciente)

classificacao = "Positivo" if res[0] == 1 else "Negativo"
print("Resultado do Tratamento:", classificacao)
print("Probabilidade:", prob)
```
